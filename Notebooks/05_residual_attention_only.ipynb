# 05 — Residual Connection after Attention Only (MiniChefGPT, Version 2)
Module: `src/residual_attention_only.py` 
Data: `recipes_clean.csv`

Design: one residual connection per block, around masked attention only:
`x = x + Attention(LN(x))`, then `x = FFN(LN(x))` with no skip.

This notebook also compares three setups so the team can decide with evidence:
`attn_only` (this version), `both` (version 1: residual after attention and FFN), and `none`.
Parts marked STAND-IN are placeholders for teammates' modules.

```python
import sys, ast
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name.lower() == "notebooks" else Path.cwd()
sys.path.append(str(ROOT))

from src.residual_attention_only import ResidualConnection

torch.manual_seed(42)
np.random.seed(42)
print("torch", torch.__version__, "| root:", ROOT)
```

```python
candidates = [ROOT / "data" / "recipes_clean.csv", ROOT / "recipes_clean.csv"]
DATA_PATH = next((p for p in candidates if p.exists()), None)
if DATA_PATH is None:
    raise FileNotFoundError("recipes_clean.csv not found. Put it in data/ or the repo root.")

df = pd.read_csv(DATA_PATH)
print(DATA_PATH.name, df.shape, list(df.columns))
df.head(3)
```

```python
def to_list(cell):
    try:
        v = ast.literal_eval(cell)
        return v if isinstance(v, list) else [str(v)]
    except (ValueError, SyntaxError):
        return [str(cell)]

def format_recipe(row):
    ingredients = ", ".join(to_list(row["ingredients"]))
    steps = " ".join(to_list(row["steps"]))
    return f"Title: {row['title']}\nIngredients: {ingredients}\nInstructions: {steps}"

print(format_recipe(df.iloc[0])[:500])
```

```python
# STAND-IN word-level tokenizer (replace with the team's tokenizer/embeddings later)
from collections import Counter

texts = [format_recipe(r) for _, r in df.head(500).iterrows()]
words = [w for t in texts for w in t.lower().split()]
common = [w for w, _ in Counter(words).most_common(1999)]
stoi = {w: i + 1 for i, w in enumerate(common)}   # 0 = <unk>
vocab_size = len(stoi) + 1
stream = torch.tensor([stoi.get(w, 0) for w in words], dtype=torch.long)
print("vocab size:", vocab_size, "| tokens:", len(stream))

D_MODEL, N_HEADS, SEQ_LEN, BATCH = 64, 4, 32, 32

def get_batch():
    ix = torch.randint(0, len(stream) - SEQ_LEN - 1, (BATCH,))
    xb = torch.stack([stream[i:i + SEQ_LEN] for i in ix])
    yb = torch.stack([stream[i + 1:i + SEQ_LEN + 1] for i in ix])
    return xb, yb

xb, yb = get_batch()
print("token ids:", tuple(xb.shape), "| targets:", tuple(yb.shape))
```

```python
B, T, D = 4, SEQ_LEN, D_MODEL
x = torch.randn(B, T, D)
sub = torch.randn(B, T, D)
res = ResidualConnection(dropout=0.0)

out = res(x, sub)
assert out.shape == x.shape
print("(a) shape preserved:", tuple(out.shape))

try:
    res(x, torch.randn(B, T, D + 1))
    raise AssertionError("mismatch was not caught")
except ValueError as e:
    print("(b) mismatch caught:", e)

assert torch.allclose(out, x + sub)
print("(c) output == x + sublayer_output")

assert torch.equal(res(x, torch.zeros_like(x)), x)
print("(d) zero sub-layer returns x unchanged")

res_d = ResidualConnection(dropout=0.5)
res_d.eval()
assert torch.allclose(res_d(x, sub), x + sub)
print("dropout is off in eval mode")
```

```python
class DemoBlock(nn.Module):
    # STAND-IN attention/FFN/LayerNorm; only ResidualConnection is mine.
    # mode = "attn_only" -> residual around attention only (VERSION 2, this notebook)
    # mode = "both"      -> residual around attention and FFN (version 1, for comparison)
    # mode = "none"      -> no residuals (for comparison)
    def __init__(self, d_model, n_heads, mode="attn_only", dropout=0.1):
        super().__init__()
        assert mode in ("attn_only", "both", "none")
        self.mode = mode
        self.ln1 = nn.LayerNorm(d_model)
        self.ln2 = nn.LayerNorm(d_model)
        self.attn = nn.MultiheadAttention(d_model, n_heads, dropout=dropout, batch_first=True)
        self.ffn = nn.Sequential(nn.Linear(d_model, 4 * d_model), nn.GELU(), nn.Linear(4 * d_model, d_model))
        self.residual1 = ResidualConnection(dropout)   # around attention
        self.residual2 = ResidualConnection(dropout)   # around FFN (used only in mode="both")

    def forward(self, x, causal_mask):
        h = self.ln1(x)
        a, _ = self.attn(h, h, h, attn_mask=causal_mask, need_weights=False)
        x = a if self.mode == "none" else self.residual1(x, a)

        f = self.ffn(self.ln2(x))
        x = self.residual2(x, f) if self.mode == "both" else f
        return x


class DemoLM(nn.Module):
    def __init__(self, vocab_size, d_model, n_heads, n_layers, mode="attn_only", dropout=0.1):
        super().__init__()
        self.tok = nn.Embedding(vocab_size, d_model)
        self.pos = nn.Embedding(SEQ_LEN, d_model)
        self.blocks = nn.ModuleList(
            [DemoBlock(d_model, n_heads, mode, dropout) for _ in range(n_layers)]
        )
        self.ln_f = nn.LayerNorm(d_model)
        self.head = nn.Linear(d_model, vocab_size)

    def forward(self, ids):
        T = ids.shape[1]
        x = self.tok(ids) + self.pos(torch.arange(T, device=ids.device))
        mask = torch.triu(torch.ones(T, T, dtype=torch.bool, device=ids.device), diagonal=1)
        for blk in self.blocks:
            x = blk(x, mask)
        return self.head(self.ln_f(x))


m = DemoLM(vocab_size, D_MODEL, N_HEADS, n_layers=2, mode="attn_only")
logits = m(xb)
assert logits.shape == (BATCH, SEQ_LEN, vocab_size)
print("2-layer attention-only model output:", tuple(logits.shape), "(batch, seq_len, vocab)")
```

```python
def grad_per_layer(mode, n_layers=12):
    torch.manual_seed(0)
    model = DemoLM(vocab_size, D_MODEL, N_HEADS, n_layers, mode, dropout=0.0)
    model.train()
    xb_, yb_ = get_batch()
    loss = nn.functional.cross_entropy(model(xb_).reshape(-1, vocab_size), yb_.reshape(-1))
    loss.backward()
    return [blk.ffn[0].weight.grad.norm().item() for blk in model.blocks]

grads = {mode: grad_per_layer(mode) for mode in ("attn_only", "both", "none")}

plt.figure(figsize=(7, 4))
for mode, g in grads.items():
    plt.plot(range(1, 13), g, "o-", label=mode)
plt.yscale("log")
plt.xlabel("Block index (1 = closest to input)")
plt.ylabel("Gradient norm (log scale)")
plt.title("Gradient reaching each block")
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

for mode, g in grads.items():
    print(f"{mode:10s} block 1 gradient: {g[0]:.5f}")
```

```python
def train(mode, n_layers=8, steps=300, lr=3e-3):
    torch.manual_seed(1)
    model = DemoLM(vocab_size, D_MODEL, N_HEADS, n_layers, mode, dropout=0.1)
    opt = torch.optim.AdamW(model.parameters(), lr=lr)
    losses = []
    model.train()
    for step in range(steps):
        xb_, yb_ = get_batch()
        loss = nn.functional.cross_entropy(model(xb_).reshape(-1, vocab_size), yb_.reshape(-1))
        opt.zero_grad(); loss.backward(); opt.step()
        losses.append(loss.item())
    return losses

def smooth(v, k=15):
    return np.convolve(v, np.ones(k) / k, mode="valid")

results = {mode: train(mode) for mode in ("attn_only", "both", "none")}

plt.figure(figsize=(7, 4))
for mode, losses in results.items():
    plt.plot(smooth(losses), label=mode)
plt.xlabel("Training step"); plt.ylabel("Loss (smoothed)")
plt.title("Next-token loss: attention-only vs both vs none")
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()

for mode, losses in results.items():
    print(f"{mode:10s} final loss (avg last 20 steps): {np.mean(losses[-20:]):.3f}")
```

## Explanation (Version 2: residual after attention only)

**Design:** one residual connection per block, around masked attention: `x = x + Attention(LN(x))`. The FFN output is passed on without a skip: `x = FFN(LN(x))`.

**Role of the residual after attention:** it keeps the embedding (which word, which position) alongside what attention adds, so attention only learns a correction.

**Gradient results** (12-layer stand-in model, one backward pass):
- Gradient norm at block 1: attention-only **0.249**, both residuals **0.048**, none **0.291**.
- With residuals around both sub-layers, the gradient stays small and flat across all 12 blocks (about 0.04–0.05).
- With attention-only residuals and with no residuals, the gradient is larger and grows towards the later blocks (roughly 0.25 up to about 0.7), and the two curves look similar.
- Gradient size alone does not show whether a model learns well, so the training loss below is the main evidence.

**Training results** (8-layer stand-in model, real recipe tokens, same seed, final loss = average of last 20 steps):
- attention-only residual: **6.083**
- residuals after attention and FFN (version 1): **3.968**
- no residuals: **6.088**

**Interpretation:** the attention-only version stalled at about the same loss as the no-residual version (about 6.1 from early in training), while the version with residuals around both attention and the FFN kept improving to about 3.97. In this experiment, the residual around the FFN was needed for the model to keep learning. A likely reason is that without a skip around the FFN, each block's output replaces `x` entirely, so the skip path is broken in every block of the 8-layer stack.

**Caveats:** short CPU run (300 steps, one seed) with stand-in blocks, so it shows the trend rather than final model quality. The stand-in blocks use LayerNorm before each sub-layer. The final design is decided by the team after the instructor's confirmation.

**Dimension rule:** the sub-layer output must match its input shape `(batch, seq_len, d_model)`, otherwise `ResidualConnection` raises a `ValueError`.
