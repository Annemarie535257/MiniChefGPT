# Position-wise Feed-Forward Network

A Transformer feed-forward network applies the same two-layer MLP to each token independently. The hidden layer expands the model dimension, GELU adds a non-linearity, and the second linear layer projects back to `d_model`. Leading dimensions (such as batch and sequence length) are preserved.

```python
import torch

from src.feed_forward import FeedForward
```

```python
batch_size = 2
sequence_length = 5
d_model = 16
d_ff = 64

feed_forward = FeedForward(d_model=d_model, d_ff=d_ff)
tokens = torch.randn(batch_size, sequence_length, d_model)
transformed = feed_forward(tokens)

print(f"Input shape:  {tuple(tokens.shape)}")
print(f"Output shape: {tuple(transformed.shape)}")
assert transformed.shape == tokens.shape
assert not torch.equal(transformed, tokens)
```

The output keeps the same batch size, sequence length, and model width while changing token representations. `FeedForward` is a standalone `nn.Module` that can be reused inside each Transformer block.
